```python
# 2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle  from it and implement the area() method in both.

# Answers:

from abc import ABC, abstractmethod
import math

# Abstract base class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

# Derived class: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Derived class: Rectangle
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle area:", circle.area())
print("Rectangle area:", rectangle.area())
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class Bank Account with private attributes  balance and methods to deposit, withdraw, and check balance?
# Answers:
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: ${amount}")
        else:
            print("Insufficient balance or invalid amount.")

    def check_balance(self):
        print(f"Current Balance: ${self.__balance}")

# Example usage
account = BankAccount(100)

account.check_balance()
account.deposit(50)
account.withdraw(30)
account.check_balance()

```

```python
# 7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static method subtract_numbers() to subtract two numbers.
# Answers:

class MathOperations:

    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

# Example usage
sum_result = MathOperations.add_numbers(10, 5)
diff_result = MathOperations.subtract_numbers(10, 5)

print("Sum:", sum_result)        # Output: Sum: 15
print("Difference:", diff_result)  # Output: Difference: 5
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two vectors?
# Answers:
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Example usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2  # This uses the overloaded + operator

print(v1)
print(v2)
print(v3)
```

```python
# 12. Implement a class Student with attributes name and grades. Create a method average_grade() to compute the average of the grades.?
# Answers:

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades  # Expecting a list of grades

    def average_grade(self):
        if len(self.grades) > 0:
            return sum(self.grades) / len(self.grades)
        else:
            return 0  # Avoid division by zero if no grades are provided

# Example usage
student1 = Student("Alice", [85, 90, 88, 92])
student2 = Student("Bob", [78, 80, 84, 76, 82])

print(f"{student1.name}'s average grade: {student1.average_grade()}")
print(f"{student2.name}'s average grade: {student2.average_grade()}")
```

```python
# 14.Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

# Answers:
# Base class: Employee
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

# Derived class: Manager
class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

# Example usage
employee = Employee("Alice", 160, 20)
manager = Manager("Bob", 160, 25, 500)

print(f"Employee Salary: ${employee.calculate_salary()}")
print(f"Manager Salary: ${manager.calculate_salary()}")
```

```python
# 25.  What is the purpose of the __call__ method in Python?
# Answers: The __call__ method in Python allows an instance of a class to be called like a function. When you implement __call__, an object can be invoked with parentheses ()
# as if it were a regular function, enabling function-like behavior for class instances.
# This is useful in scenarios like implementing callback functions, event handlers, or decorators.
# The __call__ method receives any arguments passed during the function call and can execute custom logic.
# It provides flexibility, enabling objects to act as callable entities while maintaining the structure of a class.
```
